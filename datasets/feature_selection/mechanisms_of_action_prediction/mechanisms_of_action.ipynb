### Dataset and Task Metadata

```python
from data_foundry.schema import DatasetMetadata, PredictiveMLTaskMetadata

dataset_mold = DatasetMetadata(
    unique_name="mechanisms_of_action",
    dataset_year="2021",
    domain_str="biology & life sciences",
    # Data Source
    dataset_source="Kaggle",
    original_dataset_source_download_link="https://www.kaggle.com/competitions/lish-moa/data",
    download_description="""
We download the data from the UCI repository and uzip it to a predefined folder.

mkdir -p local-data-warehouse/mechanisms_of_action/ && cd local-data-warehouse/mechanisms_of_action/ && kaggle competitions download -c lish-moa && cd ../../ && unzip local-data-warehouse/mechanisms_of_action/lish-moa.zip -d local-data-warehouse/mechanisms_of_action/ && rm local-data-warehouse/mechanisms_of_action/lish-moa.zip && rm local-data-warehouse/mechanisms_of_action/sample_submission.csv
""",
    # References
    academic_reference_bibtex="""@misc{lish2021moa,
    author = {inversion and Jin Paik and Maggie and mrbhbs and Steven Randazzo  and tnat1031},
    title = {Mechanisms of Action (MoA) Prediction},
    year = {2020},
    howpublished = {url{https://kaggle.com/competitions/lish-moa}},
    note = {Kaggle}
}
""",
    academic_reference_bibtex_key="lish2021moa",
    license="Competition license",
    data_tags=["IID"],
    curation_comments="""
- We chose the train_features and the train_targets_scored as the full dataset, because for the rest of the data, there is no target
- There are 207 targets, we decide to use the nfkb_inhibitor as the target because of its high prevalence and because it is a central inflammation/immunomodulation target.
""",
)
task_mold = PredictiveMLTaskMetadata(
    target_column_name="nfkb_inhibitor",
    problem_type="binary_classification",
    objective_metric_name="roc_auc",
    stratify_on="nfkb_inhibitor",
)
```

## Preprocessing

```python
import pandas as pd

train_features = pd.read_csv(f"{dataset_mold.path}/train_features.csv", header=0)
train_targets_scored = pd.read_csv(f"{dataset_mold.path}/train_targets_scored.csv", header=0)

df = pd.concat([train_features, train_targets_scored], axis=1)

feature_names = df.columns

cat_features = [
    "cp_type",
    "cp_time",
    "cp_dose"
]

df.drop(columns=["sig_id"], inplace=True)

df = df.sample(frac=1, random_state=42).reset_index(drop=True)

df[cat_features] = df[cat_features].astype("category")
```

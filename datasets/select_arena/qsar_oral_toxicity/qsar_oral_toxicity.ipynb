### Dataset and Task Metadata

```python
from data_foundry.schema import DatasetMetadata, PredictiveMLTaskMetadata

dataset_mold = DatasetMetadata(
    unique_name="qsar_oral_toxicity",
    dataset_year="2019",
    domain_str="biology & life sciences",
    # Data Source
    dataset_source="UCI",
    original_dataset_source_download_link="https://archive.ics.uci.edu/dataset/508/qsar+oral+toxicity",
    download_description="""
We download the data from the UCI repository and uzip it to a predefined folder. 

mkdir -p local-data-warehouse/qsar_oral_toxicity && \
wget -P local-data-warehouse/qsar_oral_toxicity/ https://archive.ics.uci.edu/static/public/508/qsar+oral+toxicity.zip && \
unzip local-data-warehouse/qsar_oral_toxicity/qsar+oral+toxicity.zip -d local-data-warehouse/qsar_oral_toxicity/ && \
rm local-data-warehouse/qsar_oral_toxicity/qsar+oral+toxicity.zip
""",
    # References
academic_reference_bibtex="""@misc{uci2019qsaroral,
    title        = {QSAR oral toxicity},
    year         = {2019},
    author       = {Unknown},
    howpublished = {UCI Machine Learning Repository},
    url = {https://doi.org/10.24432/C5PS4J},
    }
""",
    academic_reference_bibtex_key="uci2019qsaroral",
    license="CC BY 4.0",
    data_tags=["IID"],
    curation_comments="""
    - We map the original target values to "Yes" and "No" for better readability.
""",
)
task_mold = PredictiveMLTaskMetadata(
    target_column_name="Toxic",
    problem_type="binary_classification",
    objective_metric_name="roc_auc",
    stratify_on="Toxic",
)
```

## Preprocessing

```python
import pandas as pd

df = pd.read_csv(dataset_mold.path / "qsar_oral_toxicity.csv", header=None, sep=";")

target_feature = "Toxic"
df.columns = [f"Attribute {i+1}" for i in range(len(df.columns) - 1)] + [target_feature]
df[target_feature] = df[target_feature].map({"positive": "Yes", "negative": "No"})

# convert to binary type
cat_features = [col for col in df.columns]
df[cat_features] = df[cat_features].astype("category")
```

## Data Checks

```python
from data_foundry import dataset_checks
df_head, summary, numeric_stats, cat_stats, target_df = dataset_checks.run_all_checks(
    data=df,
    classification=task_mold.is_classification,
    target_feature=task_mold.target_column_name,
    print_report=False, # In notebook...
)
```
